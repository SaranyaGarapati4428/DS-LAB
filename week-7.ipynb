```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris


iris = load_iris()


df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target


plt.figure(figsize=(6, 4))
sns.histplot(df['sepal length (cm)'], bins=20, kde=True)
plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.show()

# Boxplot of Sepal Length
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['sepal length (cm)'])
plt.title("Boxplot of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.show()

# Species distribution
species_counts = df['species'].value_counts().sort_index()

plt.figure(figsize=(6, 6))
plt.pie(species_counts,labels=iris.target_names,autopct='%1.1f%%')
plt.title("Species Distribution (Pie Chart)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'],y=df['sepal width (cm)'],hue=df['species'])
plt.title("Scatterplot: Sepal Length vs Sepal Width")
plt.show()

plt.figure(figsize=(6,4))
plt.plot(df['sepal length (cm)'])
plt.title("Line chart of Sepal Length")
plt.xlabel("Sample Index")
plt.ylabel("Sepal Length (cm)")
plt.show()

plt.figure(figsize=(6,4))
sns.barplot(x=df['species'],y=df['sepal length (cm)'])
plt.title("bar chart: Mean Sepal Length by species")
plt.xticks([0,1,2],iris.target_names)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Heatmap of feature correlations
plt.figure(figsize=(8, 6))
sns.heatmap(
    df.iloc[:, :4].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Heatmap of Feature Correlations")
plt.show()


# 2. Bubble chart: Sepal Length vs Sepal Width
plt.figure(figsize=(6, 4))

plt.scatter(
    df['sepal length (cm)'],
    df['sepal width (cm)'],
    s=df['petal length (cm)'] * 20,
    alpha=0.5,
    c=df['species']
)

plt.title("Bubble Chart: Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.show()


# 3. Pair plot of Iris features
sns.pairplot(df.iloc[:, :4])
plt.suptitle("Pair Plot of Iris Features", y=1.02)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.violinplot(x=df['species'],y=df['sepal length (cm)'])
plt.title("Violin Plot: sepal length by species")
plt.xticks([0,1,2],iris.target_names)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


# Load Tips dataset
df = sns.load_dataset("tips")


# Histogram of Total Bill
plt.figure(figsize=(6, 4))
sns.histplot(df['total_bill'], bins=20, kde=True)
plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()


# Boxplot of Total Bill
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['total_bill'])
plt.title("Boxplot of Total Bill")
plt.xlabel("Total Bill")
plt.show()


# Day distribution
day_counts = df['day'].value_counts().reindex(["Thur", "Fri", "Sat", "Sun"])

plt.figure(figsize=(6, 6))
plt.pie(
    day_counts,
    labels=day_counts.index,
    autopct='%1.1f%%'
)
plt.title("Day Distribution (Pie Chart)")
plt.show()
```

```python
# Scatterplot: Total Bill vs Tip
plt.figure(figsize=(6, 4))
sns.scatterplot(
    x=df['total_bill'],
    y=df['tip'],
    hue=df['day']
)
plt.title("Scatterplot: Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()


# Line chart of Total Bill
plt.figure(figsize=(6, 4))
plt.plot(df['total_bill'])
plt.title("Line Chart of Total Bill")
plt.xlabel("Sample Index")
plt.ylabel("Total Bill")
plt.show()


# Bar chart: Mean Total Bill by Day
plt.figure(figsize=(6, 4))
sns.barplot(
    x=df['day'],
    y=df['total_bill']
)
plt.title("Bar Chart: Mean Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Mean Total Bill")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Heatmap of numerical feature correlations
plt.figure(figsize=(8, 6))

sns.heatmap(
    df[['total_bill', 'tip', 'size']].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Heatmap of Tips Dataset Correlations")
plt.show()


# 2. Bubble chart: Total Bill vs Tip
plt.figure(figsize=(6, 4))

plt.scatter(
    df['total_bill'],
    df['tip'],
    s=df['size'] * 30,
    alpha=0.5,
    c=df['size']
)

plt.title("Bubble Chart: Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()


# 3. Pair plot of Tips numerical features
sns.pairplot(
    df[['total_bill', 'tip', 'size']]
)

plt.suptitle("Pair Plot of Tips Features", y=1.02)
plt.show()
```
