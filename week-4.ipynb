```python
import pandas as pd
import seaborn as sns

# Example: load Titanic dataset from seaborn
df = sns.load_dataset("titanic")

# Fill missing values
df['age'].fillna(df['age'].median(), inplace=True)
df['embarked'].fillna(df['embarked'].mode()[0], inplace=True)

# Drop duplicates
df.drop_duplicates(inplace=True)

# Create dummy variables
df = pd.get_dummies(df, columns=['sex', 'class', 'embarked'], drop_first=True)

# Add family_size column
df['family_size'] = df['sibsp'] + df['parch']

print(df.head())
```

```python
import pandas as pd

# Load your dataset (make sure the path matches exactly)
df = pd.read_csv("student.csv")

# Fill missing numeric values with median
for col in ['Age','Attendance','Math_Score','Science_Score']:
    df[col].fillna(df[col].median(), inplace=True)

# Fill missing categorical values with mode
for col in ['Gender','Department','City']:
    df[col].fillna(df[col].mode()[0], inplace=True)

# Drop duplicates
df.drop_duplicates(inplace=True)

# Create dummy variables
df = pd.get_dummies(df, columns=['Gender','Department','City'], drop_first=True)

print(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Age Distribution')

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load your dataset (make sure the file name matches exactly)
df = pd.read_csv("student.csv")

# Compute correlation matrix (numeric columns only)
corr = df.corr(numeric_only=True)

# Print correlation values
print(corr)

# Plot heatmap
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```
